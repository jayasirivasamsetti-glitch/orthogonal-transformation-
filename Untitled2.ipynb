```python
# SCREEN ROTATION AS ORTHOGONAL TRANSFORMATION
# Google Colab - Single Cell Version

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, clear_output
import ipywidgets as widgets

# -----------------------------
# Rotation calculation
# -----------------------------
def calculate_rotation(angle):
    theta = np.radians(angle)

    Q = np.array([
        [np.cos(theta), -np.sin(theta)],
        [np.sin(theta),  np.cos(theta)]
    ])

    # Screen rectangle
    points = np.array([
        [0, 0],
        [4, 0],
        [4, 7],
        [0, 7],
        [0, 0]
    ])

    rotated = points @ Q.T

    QTQ = Q.T @ Q
    determinant = np.linalg.det(Q)

    return Q, points, rotated, QTQ, determinant


# -----------------------------
# Display function
# -----------------------------
def show_rotation(angle):

    Q, original, rotated, QTQ, determinant = calculate_rotation(angle)

    clear_output(wait=True)

    print("=" * 55)
    print("   SCREEN ROTATION AS ORTHOGONAL TRANSFORMATION")
    print("=" * 55)

    print(f"\nRotation Angle: {angle}°")

    print("\nRotation Matrix Q:")
    print(np.round(Q, 3))

    print("\nQᵀQ:")
    print(np.round(QTQ, 3))

    print("\nIdentity Matrix I:")
    print(np.eye(2))

    print("\nDeterminant:")
    print(round(determinant, 3))

    if np.allclose(QTQ, np.eye(2)):
        print("\n✓ QᵀQ = I")
        print("✓ Matrix is ORTHOGONAL")
        print("✓ Distance is preserved")

    # -----------------------------
    # Graph
    # -----------------------------
    plt.figure(figsize=(7, 6))

    plt.plot(
        original[:, 0],
        original[:, 1],
        marker="o",
        label="Original Screen"
    )

    plt.plot(
        rotated[:, 0],
        rotated[:, 1],
        marker="o",
        label="Rotated Screen"
    )

    plt.axhline(0)
    plt.axvline(0)

    plt.xlabel("X")
    plt.ylabel("Y")
    plt.title(f"Screen Rotation: {angle}°")

    plt.grid(True)
    plt.axis("equal")
    plt.legend()

    plt.show()


# -----------------------------
# Interactive buttons
# -----------------------------
button90 = widgets.Button(description="Rotate 90°")
button180 = widgets.Button(description="Rotate 180°")
button270 = widgets.Button(description="Rotate 270°")
reset = widgets.Button(description="Reset")

def show90(b):
    show_rotation(90)

def show180(b):
    show_rotation(180)

def show270(b):
    show_rotation(270)

def showreset(b):
    show_rotation(0)

button90.on_click(show90)
button180.on_click(show180)
button270.on_click(show270)
reset.on_click(showreset)

display(
    widgets.HBox([
        button90,
        button180,
        button270,
        reset
    ])
)

show_rotation(0)
```
